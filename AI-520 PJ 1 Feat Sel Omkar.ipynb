# Feature Engineering:

#### Extracting the date & time information:

Each hour is important, but each time component is all in 1 column 'DATE' so it will be more meaningful to put create separate values for each time component: HOUR, DATE, etc.

````python
# convert to datetime
df['DATE'] = pd.to_datetime(df['DATE'])

# extract components
df['year'] = df['DATE'].dt.year
df['month'] = df['DATE'].dt.month
df['day'] = df['DATE'].dt.day

# round to nearest hour
df['DATE_rounded'] = df['DATE'].dt.round('h')

# Extract just the hour (after rounding)
df['hour'] = df['DATE_rounded'].dt.hour
````

# Fixing incorrect datatypes of columns

````python
float_cols = [
    'HourlyAltimeterSetting', 'HourlyDewPointTemperature', 'HourlyDryBulbTemperature',
    'HourlyPrecipitation', 'HourlyRelativeHumidity', 'HourlySeaLevelPressure',
    'HourlyStationPressure', 'HourlyVisibility', 'HourlyWetBulbTemperature',
    'HourlyWindDirection', 'HourlyWindSpeed'
]

df[float_cols] = df[float_cols].apply(pd.to_numeric, errors='coerce')

# created a date time object as a column (called ds) for easy inputting into models created by the prophet package
df['ds'] = pd.to_datetime(df[['year', 'month', 'day', 'hour']])

df.dtypes
````

#### Filling in missing values:

````python
# missing values in each column
print("missing values per column:")
df.isna().sum()
````

````python
# fill missing values differently based on data types of columns
for col in df.columns:
    if df[col].isna().sum() > 0:  # only process columns with NaNs
        
        if np.issubdtype(df[col].dtype, np.floating):
            df[col] = df[col].fillna(df[col].mean()) # mean for float values
            
        elif np.issubdtype(df[col].dtype, np.integer):
            df[col] = df[col].fillna(df[col].median()) # median for int values
            
        else:  # object, category, string, etc.
            mode_val = df[col].mode() # mode for categorical values
            if not mode_val.empty:
                df[col] = df[col].fillna(mode_val[0])
````

````python
# verify results
print("\nRemaining missing values:")
print(df.isna().sum().sum())  # total number of NaNs left
````

# Handling duplicate date values:

For Prophet (and time-series modeling in general), each timestamp (ds) should be unique. Duplicate timestamps will confuse Prophet’s fitting process, cause errors like “non-unique ds values”, or distort trend and seasonality estimates.

It made more sense to take the average of duplicate values for each hour.

````python
# check if there are no duplicate timestamps in 'ds'

dup_vals = df[df['ds'].duplicated(keep=False)]
print(f"Number of duplicate timestamps: {dup_vals['ds'].nunique()}")
````

````python
dup_vals.head()
````

```
df = df.groupby('ds', as_index=False).mean(numeric_only=True)
```

Groups the Data Frame by the ds column (datetime). For duplicate timestamps, it finds the mean of all numeric columns ignoring non-numeric ones. It keeps only one row per unique ds.

````python
non_num = df[['ds', 'SOURCE', 'REPORT_TYPE']].drop_duplicates('ds')
num_avg = df.groupby('ds', as_index=False).mean(numeric_only=True)
df = num_avg.merge(non_num, on='ds', how='left')
````

````python
# verify there are no more duplicate timestamps in 'ds'

dup_vals = df[df['ds'].duplicated(keep=False)]
print(f"Number of duplicate timestamps: {dup_vals['ds'].nunique()}")
````
